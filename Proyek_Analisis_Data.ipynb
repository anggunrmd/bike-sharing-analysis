## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

- **Pertanyaan 1:** Pada jam berapa jumlah penyewaan sepeda paling tinggi selama periode 2011–2012?
- **Pertanyaan 2:** Bagaimana perbedaan rata-rata jumlah penyewaan sepeda pada setiap musim selama periode 2011–2012?

## Import Semua Packages/Library yang Digunakan

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Data Wrangling

### Gathering Data

Pada tahap ini, dataset yang digunakan adalah Bike Sharing Dataset. Dataset yang digunakan dalam analisis ini adalah `hour.csv` karena data tersebut menyediakan informasi jumlah penyewaan sepeda berdasarkan jam.

Dataset mencakup data penyewaan sepeda selama periode 2011–2012 dengan total 17.379 data dan 17 kolom.

#### Load df ...

```python
df = pd.read_csv('hour.csv')

df.head()
```

### Assessing Data

#### Identifying ... problem

```python
df.info()

df.duplicated().sum()

print("Rentang jam:", df['hr'].min(), "-", df['hr'].max())
print("Nilai season:", sorted(df['season'].unique()))
print("Rentang jumlah penyewaan:", df['cnt'].min(), "-", df['cnt'].max())
```

**Steps to Take:**

- Mengubah tipe data pada kolom `dteday` dari `object` menjadi `datetime` agar data tanggal dapat diproses dengan tepat.
- Menambahkan nama musim berdasarkan kode pada kolom `season` agar informasi musim lebih mudah dipahami pada tahap analisis dan visualisasi.

### Cleaning Data

Pada tahap ini dilakukan pembersihan dan penyesuaian data berdasarkan hasil assessing. Perbaikan yang dilakukan adalah mengubah tipe data pada kolom `dteday` menjadi `datetime` dan memberikan keterangan nama musim pada kolom `season`.

#### Fixing ... problem

```python
# Membuat salinan data
df_clean = df.copy()

# Mengubah tipe data dteday menjadi datetime
df_clean['dteday'] = pd.to_datetime(df_clean['dteday'])

# Menambahkan nama musim berdasarkan kode season
season_names = {
    1: 'Winter',
    2: 'Spring',
    3: 'Summer',
    4: 'Fall'
}

df_clean['season_name'] = df_clean['season'].map(season_names)

df_clean.head()
```

## Exploratory Data Analysis (EDA)

### Pertanyaan 1: Pada jam berapa jumlah penyewaan sepeda paling tinggi selama periode 2011–2012?

Untuk menjawab pertanyaan ini, dilakukan pengelompokan data berdasarkan jam (`hr`) dan menghitung total jumlah penyewaan sepeda (`cnt`) pada setiap jam.

```python
rental_by_hour = df_clean.groupby('hr')['cnt'].sum().reset_index()

rental_by_hour

max_hour = rental_by_hour.loc[rental_by_hour['cnt'].idxmax()]

print("Jam dengan jumlah penyewaan tertinggi:", max_hour['hr'])
print("Total penyewaan:", max_hour['cnt'])
```

### Pertanyaan 1:

Visualisasi berikut digunakan untuk melihat pola jumlah penyewaan sepeda pada setiap jam selama periode 2011–2012.

```python
plt.figure(figsize=(10, 5))

plt.bar(rental_by_hour['hr'], rental_by_hour['cnt'])

plt.title('Total Penyewaan Sepeda Berdasarkan Jam')
plt.xlabel('Jam')
plt.ylabel('Total Penyewaan')
plt.xticks(range(24))

plt.show()
```

### Conclusion Pertanyaan 1

Berdasarkan hasil analisis, jumlah penyewaan sepeda tertinggi terjadi pada pukul 17.00 dengan total 336.860 penyewaan selama periode 2011–2012. Jumlah penyewaan juga relatif tinggi pada pukul 18.00 dengan total 309.772 penyewaan dan pukul 08.00 dengan total 261.001 penyewaan.

### Pertanyaan 2: Bagaimana perbedaan rata-rata jumlah penyewaan sepeda pada setiap musim selama periode 2011–2012?

Untuk menjawab pertanyaan ini, dilakukan pengelompokan data berdasarkan musim (`season`) dan menghitung rata-rata jumlah penyewaan sepeda (`cnt`) pada setiap musim.

```python
rental_by_season = df_clean.groupby('season')['cnt'].mean().reset_index()

season_names = {
    1: 'Winter',
    2: 'Spring',
    3: 'Summer',
    4: 'Fall'
}

rental_by_season['season_name'] = rental_by_season['season'].map(season_names)

rental_by_season
```

```python
max_season = rental_by_season.loc[rental_by_season['cnt'].idxmax()]

print("Musim dengan rata-rata penyewaan tertinggi:", max_season['season_name'])
print("Rata-rata penyewaan:", round(max_season['cnt'], 2))
```

### Visualisasi Pertanyaan 2

Visualisasi berikut digunakan untuk melihat perbedaan rata-rata jumlah penyewaan sepeda pada setiap musim selama periode 2011–2012.

```python
plt.figure(figsize=(8, 5))

plt.bar(
    rental_by_season['season_name'],
    rental_by_season['cnt']
)

plt.title('Rata-rata Penyewaan Sepeda Berdasarkan Musim')
plt.xlabel('Musim')
plt.ylabel('Rata-rata Penyewaan')

plt.show()
```

### Conclusion Pertanyaan 2

Berdasarkan hasil analisis, rata-rata jumlah penyewaan sepeda berbeda pada setiap musim. Musim Summer memiliki rata-rata penyewaan tertinggi, yaitu sebesar 236,02 penyewaan. Sementara itu, Winter memiliki rata-rata penyewaan terendah, yaitu sebesar 111,11 penyewaan.

## Conclusion & Recommendation

- **Conclusion pertanyaan 2:** Rata-rata jumlah penyewaan sepeda berbeda pada setiap musim. Musim Summer memiliki rata-rata penyewaan tertinggi sebesar 236,02 penyewaan, sedangkan Winter memiliki rata-rata terendah sebesar 111,11 penyewaan.
